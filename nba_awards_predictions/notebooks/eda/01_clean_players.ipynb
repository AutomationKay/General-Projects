# Player data cleaning notebook

```python
import pandas as pd
import os
```

```python
RAW_PATH = "../../data/raw"
CLEANED_PATH = "../../data/cleaned"
os.makedirs(CLEANED_PATH, exist_ok=True)
```

### Step 1: Load in and preview data

```python
def load_and_player_data(start_year = 2010, end_year = 2025 ):
    player_df = []

    for year in range(start_year, end_year +1):
        file_path = f"{RAW_PATH}/nba_combined_stats_{year}.csv"
        if os.path.exists(file_path):
            df = pd.read_csv(file_path)
            df["Season"] = year
            player_df.append(df)
        else:
            print(f"Missing: {file_path}")
    
    player_merged = pd.concat(player_df, ignore_index=True)
    return player_merged
```

```python
player_merged = load_and_player_data()
```

```python
def describe_dataframe(df):
    print("  TYPES   ".center(82, '-'))
    print(df.dtypes)

    print("  SHAPE   ".center(82,'-'))
    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    print("  DUPLICATES  ".center(82,'-'))
    print(df.duplicated().sum())

    print("  NaN VALUES".center(82,'-'))
    print(df.isna().sum())

    #print("  QUANTILES   ".center(82,'-'))
    #print(df.quantile([0,0.05,0.50,0.95,0.99,1]).T)

    print("  DATAFRAME HEAD      ".center(82, '-'))
    print(df.head(n=5))
    print("  DATAFRAME TAIL      ".center(82, '-'))
    print(df.tail(n=5))
```

```python
describe_dataframe(player_merged)
```

Next we'll take a count of the unique Seasons present in the data. There should be an award winner in each of the three award categories per season. So if there are 15 seasons, there should be 15 MVPs, 15 DPOY, and 15 Scoring Leaders. If the counts don't match, this means something went wrong in the get_award_winners.py file and will have to be revisited.

```python
print(f"Number of seasons: {player_merged["Season"].nunique()}")
print(f"Count of MVPS: {player_merged["is_MVP"].sum()}")
print(f"Count of DPOY: {player_merged["is_DPOY"].sum()}")
print(f"Count of Scoring leaders: {player_merged["is_PPG_Leader"].sum()}")
```

```python
player_merged.to_csv(f"{CLEANED_PATH}/player_all_years.csv", index=False)
print(f"Saved cleaned team dataset to: {CLEANED_PATH}/player_all_years.csv")
```
